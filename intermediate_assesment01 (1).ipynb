### 2.1 Calculate the total revenue for the store (sum of all quantities × price per item).

```python
total_rev = 0
for sale in sales_data:
    total_rev = total_rev + (sale["quantity"]*sale["price_per_item"])

print("Total Revenue = ", total_rev)
```

### 2.2 Identify all orders where the total order value (quantity × price per item) exceeds ₹2,000. Store the result as a new list.

```python
order_above_2000=[]
for sale in sales_data:
    total_rev = total_rev + (sale["quantity"]*sale["price_per_item"])
    if total_rev > 2000:
        order_above_2000.append(sale)

print(order_above_2000)
```

### 2.3 Find the item with the highest total quantity sold across all orders.

```python
highest = max(sales_data, key=lambda x: x["quantity"])

print("Item:", highest["item"])
print("Quantity:", highest["quantity"])
```

### 3.1 Write a function apply_discount(orders, discount_rate) that takes a list of orders and applies a percentage discount (e.g., 10% = 0.10) to the price_per_item for each order.Return a new list of orders (do not modify the original list).

```python
def apply_discount(orders, discount_rate):
    new_orders = []

    for order in orders:
        new_order = order.copy()
        new_order["price_per_item"] = order["price_per_item"] * (1 - discount_rate)
        new_orders.append(new_order)

    return new_orders


discounted_orders = apply_discount(sales_data, 0.10)

print(discounted_orders)
```

### 4.1 Write a custom iterator class OrderIterator to iterate through the orders, yielding one order at a time.

```python
class OrderIterator:
    def __init__(self, orders):
        self.orders = iter(orders)

    def __iter__(self):
        return self

    def __next__(self):
        return next(self.orders)


order_iterator = OrderIterator(sales_data)

for order in order_iterator:
    print(order)
```

### 4.2 Create a dictionary grouping all orders by customer using a dictionary comprehension.

```python
customers = {order["customer"] for order in sales_data}

orders_by_customer = {
    customer: [order for order in sales_data if order["customer"] == customer]
    for customer in customers
}

print(orders_by_customer)
```
