## Handling Missing Values

```python
data.isnull().sum()
```

```python
data['No of Times Visited'].value_counts(dropna=False)
```

```python
data['No of Times Visited'] = data['No of Times Visited'].fillna('None')
```

```python
data['No of Times Visited'].value_counts(dropna=False)
```

```python
numerical_columns = data.select_dtypes(
    include=np.number
).columns

numerical_columns
```

```python
for col in numerical_columns:

    if col != 'Sale Price':

        if data[col].isnull().sum() > 0:

            data[col] = data[col].fillna(
                data[col].median()
            )
```

```python
categorical_columns = data.select_dtypes(
    include='object'
).columns

categorical_columns
```

```python
for col in categorical_columns:

    if data[col].isnull().sum() > 0:

        data[col] = data[col].fillna(
            data[col].mode()[0]
        )
```

```python
data.dropna(subset=['Sale Price'], inplace=True)
```

## Outlier Removal

```python
outlier_columns = [
    'No of Bedrooms',
    'No of Bathrooms',
    'Flat Area (in Sqft)',
    'Lot Area (in Sqft)',
    'Area of the House from Basement (in Sqft)',
    'Basement Area (in Sqft)',
    'Living Area after Renovation (in Sqft)',
    'Lot Area after Renovation (in Sqft)'
]
```

```python
for col in outlier_columns:

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[col] < lower_bound) |
        (data[col] > upper_bound)
    ]

    print(
        col,
        "| Lower:", round(lower_bound, 2),
        "| Upper:", round(upper_bound, 2),
        "| Outliers:", len(outliers)
    )
```

```python
for col in outlier_columns:

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[col] < lower_bound) |
        (data[col] > upper_bound)
    ]

    print("\n", col)
    print("Minimum:", data[col].min())
    print("Maximum:", data[col].max())
    print("Outlier Minimum:", outliers[col].min())
    print("Outlier Maximum:", outliers[col].max())
```

#### Outlier Treatment: Outliers were detected using the IQR method. Since extreme values in housing features may represent legitimate large or luxury properties, IQR outliers were not removed automatically. The bedroom distribution was examined individually, and a single observation containing 33 bedrooms was identified as an extreme anomaly compared with the remaining data, where the next-highest value was 11 bedrooms. Therefore, only the 33-bedroom observation was removed, while other plausible extreme observations were retained.
