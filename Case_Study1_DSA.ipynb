### Bank Acount Manager

```python
account_holder = input("Enter Account Holder Name: ")
account_number = input("Enter Account Number: ")
balance = 0.0

def deposit(amount):
    global balance
    balance += amount
    print("Amount Deposited:", amount)
    print("Current Balance:", balance)

def withdraw(amount):
    global balance
    if amount <= balance:
        balance -= amount
        print("Amount Withdrawn:", amount)
        print("Current Balance:", balance)
    else:
        print("Insufficient Balance!")

def check_balance():
    print("Account Holder:", account_holder)
    print("Account Number:", account_number)
    print("Available Balance:", balance)

while True:
    print("\n------ Bank Account Manager ------")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Balance Enquiry")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        amount = float(input("Enter amount to deposit: "))
        deposit(amount)

    elif choice == 2:
        amount = float(input("Enter amount to withdraw: "))
        withdraw(amount)

    elif choice == 3:
        check_balance()

    elif choice == 4:
        print("Thank you for using Bank Account Manager.")
        break

    else:
        print("Invalid Choice!")
```

### Weather Data Analyzer

```python
region = input("Enter Region: ")

temp1 = float(input("Enter Temperature 1: "))
temp2 = float(input("Enter Temperature 2: "))
temp3 = float(input("Enter Temperature 3: "))

hum1 = float(input("Enter Humidity 1: "))
hum2 = float(input("Enter Humidity 2: "))
hum3 = float(input("Enter Humidity 3: "))

def average_temperature():
    return (temp1 + temp2 + temp3) / 3

def average_humidity():
    return (hum1 + hum2 + hum3) / 3

def weather_condition(tempo, humy):
    if tempo > 30:
        print("Weather Condition: Sunny")
    elif humy > 70:
        print("Weather Condition: Rainy")
    else:
        print("Weather Condition: Cloudy")

avg_temp = average_temperature()
avg_hum = average_humidity()

print("\nRegion:", region)
print("Average Temperature:", avg_temp)
print("Average Humidity:", avg_hum)

weather_condition(avg_temp, avg_hum)
```
